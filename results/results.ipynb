```python
import pandas as pd
import glob

csv_files = glob.glob("*.csv")
df_list = [pd.read_csv(file) for file in csv_files]
df = pd.concat(df_list, ignore_index=True)
```

```python
df
```

```python
df = df.drop(columns=["dtype", "device", "optimize_hyper_parameters"], axis=1)
```

```python

grouped = (
    df.groupby(
        ["model", "backbone", "features", "lr", "use_embedding", "sqrt", "batch_size"],
        dropna=False
    ).mean()
)

with pd.option_context(
    "display.max_rows", None,        # show every row
    "display.max_columns", None,     # show every column
    "display.width", None,           # don’t wrap long lines
):
    display(grouped)                 # or just `grouped` in a cell
```

```python
best_nll_per_backbone = df.loc[df.groupby(["backbone", "features"])["test_nll"].idxmin()]
best_nll_per_backbone[["model", "test_nll", "test_ece", "test_auc", "backbone", "lr", "sqrt", "features", "use_embedding", "batch_size"]]
```
